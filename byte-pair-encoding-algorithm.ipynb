```python
import numpy as np
```

Initialize the vocabulary

```python
# some text with lots of repititions
text = 'like liker love lovely hug hugs hugging hearts'

chars = list(set(text))
chars.sort()

for i in chars:
    print(f"{i} apears {text.count(i)} times")
```

```python
# make the vocabulary
vocab = {word:i for i,word in enumerate(chars)}
vocab
```

Find character pairs and merge the most frequent

```python
# the text needs to be in list, not a string
# each element in the list is a token
originalText = list(text)   # This is going to change alot that's why named original text

tokenPairs = dict()

# loop over tokens
for i in range(len(originalText)-1):
    # create a pair
    pair = originalText[i] + originalText[i+1]
    # increase pair frequencies
    if pair in tokenPairs:
        tokenPairs[pair] += 1
    else:
        tokenPairs[pair] = 1
print(tokenPairs)
# for i in tokenPairs:
#     print(f"{i} : {tokenPairs[i]}")
      

```

```python
# find the most frequent pair
mostFrequentPairIndex = np.argmax(list(tokenPairs.values()))    # gives the index of the maximum value in a list or array
mostFrequentPairCharacter = list(tokenPairs.keys())[mostFrequentPairIndex]
print(f"The most frequency character pair is {mostFrequentPairCharacter} with {tokenPairs[mostFrequentPairCharacter]} appearances")
```

```python
# update the vocab
vocab[mostFrequentPairCharacter] = max(vocab.values()) + 1
vocab
```

Replace the token pair with one token

```python
# Initialize a new text list
newText = []

# loop through the list
i=0
while (i <len(originalText)-1):
    # test wether the pair of this and the following elements matches the newly created list
    if originalText[i] + originalText[i+1] == mostFrequentPairCharacter:
        # append to the new version of the text
        newText.append(mostFrequentPairCharacter)
        print(f"added: {mostFrequentPairCharacter}")
        # skip the next character
        i += 2
    else:   # this isn't a merged pair, so add this token to the list
        newText.append(originalText[i])
        i += 1
print('\n')
print(f"originalText: {originalText}\n")
print(f"newText: {newText}")
```

Create functions

```python
def getPairStatistics(textToPair):
    tokenPairs = {}
    for i in range(len(textToPair)-1):
        pair = textToPair[i]+textToPair[i+1]

        if pair in tokenPairs:
            tokenPairs[pair] += 1
        else:
            tokenPairs[pair] = 1

    return tokenPairs
# ----------------------------------------------------------

def updateVocab(tokenPairs, vocab):
    mostFrequentPairIndex = np.argmax(list(tokenPairs.values()))
    newToken = tokenPairs.keys()[mostFrequentPairIndex]
    vocab[newToken] = max(tokenPairs.values()) + 1

    return newToken, vocab

# ----------------------------------------------------------

def generateNewTokenSequence(previousText,newToken):
    newText = []
    i = 0
    while i <(len(previousText)-1):
        if((previousText[i]+previousText[i+1]) == newToken):
            newText.append(newToken)
            i+=2
        else:
            newText.append(previousText[i])
            i+=1
    return newText
```
