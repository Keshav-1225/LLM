Gpt's tokenizer

```python
tokenizer = tiktoken.get_encoding('cl100k_base')
# dir(tokenizer)
```

```python
tokenizer.n_vocab
tokenizer.decode([1014])
```

Tokenization!

```python
token = tokenizer.encode("Wake up to reality. Nothing here goes as planned in this accursed world")
print(token)
print(tokenizer.decode([tokenizer.eot_token]))

# explore some tokens
for i in range(1000,1100):
    print(f"{i} : {tokenizer.decode([i])}")
```

```python
text = "Wake up to reality. Nothing here goes as planned in this acursed world. The longer you live the more you realize that the only thing that exist in this world is mere pain suffering and futility. I am the Ghost of the Uchiha"
for word in text.split():
    print(f"word: {word}\t tokens:{tokenizer.encode(word)}")
```
