```python
import numpy as np
text = 'like liker love lovely hug hugs hugging hearts'

chars = sorted(set(text))
print("Characters: ",chars)

vocab = {w:i for i,w in enumerate(chars)}
print("Vocabulary: ",vocab)
```

```python
def getPairStatistics(textToPair):
    tokenPairs = {}
    for i in range(len(textToPair)-1):
        pair = textToPair[i]+textToPair[i+1]

        if pair in tokenPairs:
            tokenPairs[pair] += 1
        else:
            tokenPairs[pair] = 1

    return tokenPairs
# ----------------------------------------------------------

def updateVocab(tokenPairs, vocab):
    mostFrequentPairIndex = np.argmax(list(tokenPairs.values()))
    newToken = list(tokenPairs.keys())[mostFrequentPairIndex]
    vocab[newToken] = max(vocab.values()) + 1

    return newToken, vocab

# ----------------------------------------------------------

def generateNewTokenSequence(previousText,newToken):
    newText = []
    i = 0
    while i <(len(previousText)-1):
        if((previousText[i]+previousText[i+1]) == newToken):
            newText.append(newToken)
            i+=2
        else:
            newText.append(previousText[i])
            i+=1
    return newText
```

```python
originaltext = list(text)
tokenPairs = getPairStatistics(originaltext)
(newToken, vocab)=updateVocab(tokenPairs, vocab)
newText = generateNewTokenSequence(originaltext,newToken)

def tokenizer(originaltext, vocab):
    tokenPairs = getPairStatistics(originaltext)
    newToken, vocab = updateVocab(tokenPairs, vocab)
    newText = generateNewTokenSequence(originaltext, newToken)
    return newText
# print("Original text: ",originaltext)
# print("\nNew Text: ",newText)
originaltext = tokenizer(originaltext,vocab)
for i in range(8):
    originaltext = tokenizer(originaltext,vocab)
print(originaltext)

```
